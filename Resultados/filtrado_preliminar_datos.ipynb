Calculo descriptores clásicos para la artemisinina porque me faltan

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Crippen, Lipinski

smiles_artemisinina = "C[C@@H]1CC[C@H]2[C@H](C(=O)O[C@H]3[C@@]24[C@H]1CC[C@](O3)(OO4)C)C"


mol = Chem.MolFromSmiles(smiles_artemisinina)

datos = {
    "SMILES_Quiral": smiles_artemisinina,
    "MW": Descriptors.MolWt(mol),
    "LogP": Crippen.MolLogP(mol),
    "HBD": Lipinski.NumHDonors(mol),
    "HBA": Lipinski.NumHAcceptors(mol),
    "TPSA": Descriptors.TPSA(mol),
    "RotBonds": Lipinski.NumRotatableBonds(mol)
}

df_artemisinina = pd.DataFrame([datos])

df_artemisinina
```

```python
df_global_descriptor_artemisinin
```

```python
df_artemisinina_completo = pd.concat(
    [df_artemisinina.reset_index(drop=True), df_global_descriptor_artemisinin.reset_index(drop=True)],
    axis=1
)
df_artemisinina_completo
```

Ahora obtengo el df completo que incluye la artemisinina

```python
df_fase1 = pd.concat([df_artemisinina_completo, df_fase1_completo],ignore_index=True)
df_fase1
```
