```python
from rdkit import Chem

smi = "C[C@H]1CC[C@@H]2[C@@H]([C@@H](O[C@@H]3[C@]24[C@@H]1CC[C@@](O3)(OO4)C)O[*:1])C"
m = Chem.MolFromSmiles(smi)
assert m is not None

# ¿Hay calcio?
has_ca = any(a.GetSymbol() == "Ca" for a in m.GetAtoms())
print("Has Ca:", has_ca)

# Imprime símbolos únicos
print("Unique elements:", sorted(set(a.GetSymbol() for a in m.GetAtoms())))

print("iso:", Chem.MolToSmiles(m, isomericSmiles=True))
```
